# Tokenizando os dados

````python
dados = []

# Abre o arquivo ZIP e processa cada arquivo contido nele
with zipfile.ZipFile('/content/texts.zip') as zip:
    for nome_arquivo in zip.namelist():  # Itera sobre a lista de arquivos no ZIP
        with zip.open(nome_arquivo) as arquivo:  # Abre cada arquivo dentro do ZIP
            conteudo = arquivo.read().decode('utf-8')  # Lê e decodifica o conteúdo
            palavras = conteudo.split()  # Divide o conteúdo em palavras
            
            # Adiciona cada palavra junto com o nome do arquivo à lista de dados
            for palavra in palavras:
                dados.append([nome_arquivo, palavra])

# Cria um DataFrame do Pandas com as palavras extraídas e seus arquivos de origem
tabela_palavras = pd.DataFrame(dados, columns=['arquivo', 'palavra'])
````

**Rotulando os dados no formato IOB**

O formato BIO/IOB ,abreviação de inside, outside, beginning, é um formato de marcação comum para marcar tokens em uma tarefa de agrupamento em linguística computacional (por exemplo, reconhecimento de entidade nomeada).

- O prefixo B antes de uma tag indica que a tag é o início de um pedaço
- O prefixo I antes de uma tag indica que a tag está dentro de um pedaço.
- O prefixo O indica que um token não pertence a nenhuma entidade/pedaço.

Exemplo:


| **palavra** |     **label** |
|------------:|--------------:|
|     ACORDAO |             O |
|    CONSELHO | B-ORGANIZACAO |
|    ESPECIAL | I-ORGANIZACAO |
|          DF |       B-LOCAL |

````python
grupos_arquivos = tabela_palavras.groupby(by= 'arquivo')
````

````python
grupos_arquivos.get_group('ADI2TJDFT.txt')
````

````python
# Obtém os dados do grupo 'ADI2TJDFT.txt' contendo palavras e seus respectivos rótulos
tabela_agrupada = grupos_arquivos.get_group('ADI2TJDFT.txt')[['palavra', 'label']].values

# Inicializa uma string para armazenar o conteúdo do texto extraído
conteudo = ''

# Inicializa um dicionário para armazenar as anotações das entidades
anotacoes = {'entities': []}

# Variáveis para armazenar as posições de início e fim das palavras no texto
inicio_palavra = 0
fim_palavra = 0

# Itera sobre cada palavra e seu respectivo rótulo na tabela agrupada
for texto, label in tabela_agrupada:
    texto = str(texto)  # Converte a palavra para string (caso necessário)
    tamanho_texto = len(texto) + 1  # Calcula o tamanho do texto incluindo um espaço

    # Define os índices de início e fim da palavra no texto concatenado
    inicio_palavra = fim_palavra
    fim_palavra = inicio_palavra + tamanho_texto

    # Se o rótulo não for 'O', adiciona a anotação da entidade
    if label != 'O':
        anotacao = (inicio_palavra, fim_palavra - 1, label)
        anotacoes['entities'].append(anotacao)

    # Concatena a palavra ao conteúdo final, adicionando um espaço entre elas
    conteudo = conteudo + texto + ' '
````

````python
arquivos = grupos_arquivos.groups.keys()
````

````python
documentos = []

# Itera sobre cada arquivo na lista de arquivos
for arquivo in arquivos:
    documento = []  # Inicializa a estrutura do documento
    
    # Obtém a tabela de palavras e rótulos associadas ao arquivo
    tabela_agrupada = grupos_arquivos.get_group(arquivo)[['palavra', 'label']].values
    
    # Inicializa variáveis para armazenar o conteúdo e anotações do documento
    conteudo = ''
    anotacoes = {'entities': []}
    
    # Variáveis para rastrear as posições das palavras dentro do texto
    inicio_palavra = 0
    fim_palavra = 0
    
    # Itera sobre cada palavra e seu respectivo rótulo na tabela agrupada
    for texto, label in tabela_agrupada:
        texto = str(texto)  # Converte o texto para string, se necessário
        tamanho_texto = len(texto) + 1  # Calcula o tamanho da palavra incluindo o espaço
        
        # Define os índices de início e fim da palavra dentro do conteúdo total
        inicio_palavra = fim_palavra
        fim_palavra = inicio_palavra + tamanho_texto
        
        # Se a palavra possuir um rótulo diferente de 'O', registra a anotação
        if label != 'O':
            anotacao = (inicio_palavra, fim_palavra - 1, label)
            anotacoes['entities'].append(anotacao)
        
        # Concatena a palavra ao conteúdo do documento, adicionando um espaço
        conteudo = conteudo + texto + ' '
    
    # Estrutura o documento com seu conteúdo e respectivas anotações
    documento = (conteudo, anotacoes)
    
    # Adiciona o documento processado à lista de documentos
    documentos.append(documento)
````

# Construindo função de treinamento

````python
from tqdm import tqdm
from spacy.training import Example
````

````python
def treinar_modelo_ner(dados_treino, dados_validacao, epocas):
    # Carrega um modelo pré-treinado da spaCy para o idioma português
    modelo = spacy.load('pt_core_news_sm')
    
    # Verifica se o pipeline de NER já existe no modelo
    if 'ner' not in modelo.pipe_names:
        ner = modelo.create_pipe('ner')  # Cria um novo pipeline de NER
        modelo.add_pipe(ner, last=True)  # Adiciona o pipeline ao modelo
    else:
        ner = modelo.get_pipe('ner')  # Obtém o pipeline de NER existente
    
    # Adiciona rótulos das entidades ao pipeline de NER
    for _, anotacoes in dados_treino:
        for ent in anotacoes.get('entities'):
            ner.add_label(ent[2])
    
    # Desativa outros pipelines para treinar apenas o NER
    outros_pipelines = [pipeline for pipeline in modelo.pipe_names if pipeline != 'ner']
    with modelo.disable_pipes(*outros_pipelines):
        
        # Define uma semente para garantir reprodutibilidade nos resultados
        spacy.util.fix_random_seed()
        
        # Cria um otimizador para o modelo
        otimizador = modelo.create_optimizer()
        
        # Loop de treinamento por número de épocas
        for epoca in tqdm(range(epocas), desc='Treinando o modelo'):
            
            random.seed(10)  # Define semente para aleatoriedade controlada
            random.shuffle(dados_treino)  # Embaralha os dados de treinamento
            losses = {'ner': 0.0}  # Inicializa a variável de perda do modelo
            
            # Treina o modelo com os dados de treinamento
            for textos, anotacoes in dados_treino:
                exemplo = Example.from_dict(modelo.make_doc(textos), anotacoes)  # Cria um exemplo para treinamento
                modelo.update([exemplo], drop=0.2, sgd=otimizador, losses=losses)  # Atualiza o modelo
            
            # Exibe a perda média por amostra na época atual
            print(f'\nÉpoca {epoca+1} - Loss médio de treino: {losses["ner"] / len(dados_treino)}')
            
            # Inicializa a perda de validação
            val_losses = {'ner': 0.0}
            exemplos = []
            
            # Cria exemplos de validação
            for textos, anotacoes in dados_validacao:
                exemplo = Example.from_dict(modelo.make_doc(textos), anotacoes)
                exemplos.append(exemplo)
            
            # Avalia o modelo nos dados de validação
            for exemplo in exemplos:
                modelo.update([exemplo], sgd=None, drop=0, losses=val_losses)
            
            # Exibe a perda média de validação na época atual
            print(f'\nÉpoca {epoca+1} - Loss médio de validação: {val_losses["ner"] / len(dados_validacao)}')
    
    return modelo
````

**Treinando e salvando o modelo**

````python
modelo_ner = treinar_modelo_ner(dados_treino, dados_validacao, epocas = 30)
````
